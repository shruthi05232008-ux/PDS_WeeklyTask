# Data cleaning and preprocesing

```python
df["Order Date"]=pd.to_datetime(df["Order Date"])
df["Ship Date"]=pd.to_datetime(df["Ship Date"])
df[["Order Date","Ship Date"]].dtypes
```

```python
text_columns=["Category","Sub-Category","Segment"]
for column in text_columns:
    df[column]=(
        df[column]
        .astype(str)
        .str.strip()
        .str.title()
    )
```

```python
numerical_features = df.select_dtypes(include='number').columns
```

```python
summary = pd.DataFrame({
    "Mean": df[numerical_features].mean(),
    "Median": df[numerical_features].median(),
    "Minimum": df[numerical_features].min(),
    "Maximum": df[numerical_features].max(),
    "Standard Deviation": df[numerical_features].std()
})

summary
```

```python
category_profit=(
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)
category_profit
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Category",
    y="Sales",
    estimator="sum"
)

plt.title("Category-wise Total Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()

```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Category",
    y="Sales",
    estimator="sum"
)

plt.title("Category-wise Total Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    y="Sales",
)

plt.title("Sales Distribution")
plt.ylabel("Total Sales")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Category",
    y="Profit",
)

plt.title("Profit Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")


plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit",
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()
```

# Sub Category Wise profit

```python
sub_category_profit=(
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)
sub_category_profit
```

```python
df["Profit Margin"] = (df["Profit"] / df["Sales"]) * 100
df[["Sales","Profit","Profit Margin"]].head()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit Margin",
)

plt.title("Discount vs Profit ")
plt.xlabel("Discount")
plt.ylabel("Profit Margin(%)")

plt.show()
```
