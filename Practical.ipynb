# Prc 3 Program to define and call user-defined functions

```python
def greet():
    print("Hello")
def add(a,b):
    return a+b
def multiply(a,b):
    return a*b
def welcome(name="Student"):
    print("Welcome",name)
def total(*marks):
    return sum(marks)
square=lambda x:x*x
def factorial(n):
    if n<=1:
        return 1
    return n*factorial(n-1)
greet()
print("Addition:",add(20,30))
print("Multiplication:",multiply(5,3))
welcome()
welcome("Shruthi")
print("Total:",total(80,90,85))
print("Square:",square(5))
print("Factorial:",factorial(5))
    
```

#  Prc 4 List and List Comprehensions

```python
a=[1,2,3,4,5]
print("Original list:",a)
print("First:",a[0])
print("Last:",a[-1])
a.append(6)
a.insert(1,10)
a.remove(3)
print("Updated list:",a)
print("Length:",len(a))
print("Elements:")
for x in a:
    print(x)
squares=[x*x for x in a]
even=[x for x in a if x%2==0]
odd=[x for x in a if x%2!=0]
double=[x*2 for x in a]
print("Squares:",squares)
print("Even:",even)
print("Odd:",odd)
print("Double:",double)
matrix=[[1,2],[3,4]]
flat=[x for row in matrix for x in row]
print("Flattened:",flat)
    
```
